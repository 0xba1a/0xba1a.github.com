```python
from sklearn.datasets import fetch_openml

mnist = fetch_openml('mnist_784', as_frame=False)
X, y = mnist.data, mnist.target
print(f"Feature matrix shape: {X.shape}")
```

70,000 images. Each as 28 x 28 pixels (784)
Each pixel represents a gray scale value between 0 (white) and 255 (black)

```python
import matplotlib.pyplot as plt

some_digit = X[36000]
some_digit_image = some_digit.reshape(28, 28)
plt.imshow(some_digit_image, cmap='binary')
plt.axis('off')
plt.show()

print(f"Label: {y[36000]}")
```

The MINST dataset is split into training and test set by default. The first 60,000 are training whereas the rest are test set. Also, the training set is shuffled sufficiently. So, that it will perform better for "cross-validation" (Create n folds to train on n-1 folds and test on the remaining fold. Keep do it with respect to every fold)
